```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph()

usuarios = ['A', 'B', 'C', 'D', 'E']
G.add_nodes_from(usuarios)

retweets = [
    ('A', 'B'),
    ('A', 'C'),
    ('B', 'C'),
    ('C', 'D'),
    ('D', 'A'),
    ('E', 'A'),
    ('E', 'C')
]
G.add_edges_from(retweets)

in_degree_centrality = G.in_degree()

print(f"Nodos añadidos al grafo: {G.nodes()}")
print(f"Aristas (retweets) añadidas al grafo: {G.edges()}")

plt.figure(figsize=(10, 8))
pos = nx.spring_layout(G, seed=42)
nx.draw_networkx(G, pos, with_labels=True, node_color='lightgreen', node_size=2500,
                 arrowsize=25, font_size=14, font_weight='bold', width=1.5, edge_color='gray')
plt.title("Grafo de Retweets de Twitter (Retweets a -> Retweets de B)", size=16)
plt.axis('off')
plt.show()

usuarios_influyentes = sorted(in_degree_centrality, key=lambda item: item[1], reverse=True)

print("\n--- Análisis de Influencia por Retweets Recibidos (In-Degree) ---")
for usuario, retweets_recibidos in usuarios_influyentes:
    print(f"El Usuario {usuario} ha sido retuiteado {retweets_recibidos} veces.")

print(f"\nEl usuario más influyente es: {usuarios_influyentes[0][0]} con {usuarios_influyentes[0][1]} retweets recibidos.")
```
